```python
# criar variáveis  para armazenar a senha
senha = "123456"
senha_criptografada = " "

# ler os caracteres
for numero in senha:
  numero = int(numero) + 1
  senha_criptografada = senha_criptografada + str(numero)
print(senha_criptografada)
```

```python
dicionario = "qwertyuiopasdfghjklçzxcvbnm"
alfabeto = "abcdefghijklmnopqrstuvwxyz"

def codifica(senha) :
  senha_criptografada = " "
  for lecra in senha:
    if "a" <= letra <= "z":
    posicao = ord(letra) - ord("a")
    senha_criptografada += dicionario[posicao]
  else:
        senha_criptografada += letra
  return senha_criptografada

senha = "gerson123"
print(f"A senha '{senha})' criptografada é: {codifica(senha)}'
```
